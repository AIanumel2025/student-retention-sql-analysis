```python
import pandas as pd
import numpy as np
from faker import Faker
import random
from datetime import datetime, timedelta
```

```python
# dataset to predict student chances of graduation
data = []

for i in range(50):
  student = {
      "student_id": i + 1,
      "age": round(random.uniform(1,17), 0),
      "gender": random.choice(["Male", "Female"]),
      "socio_economic_score": random.choice(["Poor",
                                             "Lower_class",
                                             "Upper_middle_class",
                                             "Upper_class",
                                             "Wealthy"]),
      "attendance_rate": round(random.uniform(0, 100), 2),
      "average_assignment_score": random.randint(0,100),
      "midterm_exam_score": random.randint(0, 100),
      "final_exam_score": random.randint(0, 100),
      "study_hours_per_week": random.randint(0, 100),
      "lms_logins_per_week": random.randint(0, 100),
      "part_time_job_hours": random.randint(0, 100),
      "financial_aid": random.choice(["True", "False"]),
      "disciplinary_record": random.choice(
          [
          "None",
          "Moderate",
          "High"
      ]
          ),
      "graduation_outcome": random.choice(
          [
          "On_Time",
          "Delayed",
          "Dropout_Risk",
          "High_Achiever"
      ]
          )

  }

  data.append(student)

  df_=pd.DataFrame(data)
```
