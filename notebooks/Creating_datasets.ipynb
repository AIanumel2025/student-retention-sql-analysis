```python
import pandas as pd
import numpy as np
from faker import Faker
import random
from datetime import datetime, timedelta
```

### Creating a users table

```python
fake = Faker()

users = []
for i in range(100):
  user = {
      "user_id": i + 1,
      "name": fake.name(),
      "email": fake.email(),
      "country": fake.country(),
      "signup_date": fake.date_between(start_date='-2y', end_date='today'),
      "subscription_type": random.choice(
          ["Free", "Premium", "Pro", "Enterprise"]
      )
  }
  users.append(user)

df1=pd.DataFrame(users)
df1

```

```python
people = []
for i in range(100000):
  person = {
      "person_id": i + 1,
      "name": fake.name(),
      "address": fake.address(),
      "city": fake.city(),
      "occupation": fake.job(),
      "email": fake.email(),
      "signup_date": fake.date_between(start_date='-2y', end_date='today'),
      "phone_number": fake.phone_number()
  }

  people.append(person)

df2=pd.DataFrame(people)
df2
```

```python
df2.to_csv('occupational_data', index=False)
```

# Creating a models table

```python
models = []

model_names = [
    "Random Forest Classifier",
    "Histogram Gradient Boosting",
    "Support Vector Machines",
    "Logistic Regression",
    "BERT",
    "DistilBERT"
]

for i in range(35):
  model={
      "model_ID": i + 1,
      "model_name": random.choice(model_names),
      "accuracy": round(random.uniform(0.60,0.99),2),
      "deployment_date": fake.date_between(start_date= '-1y', end_date = 'today')
  }

  models.append(model)

df3=pd.DataFrame(models)
df3
```

# Create an Inference Log table

```python
logs = []

for i in range(25):
  log={
      "log_id": i + 1,
      "user_id": random.randint(1,100),
      "model_id": random.randint(1,20),
      "latency_ms": random.randint(50, 2000),
      "request_time": fake.date_time_between(start_date= '-1y', end_date= 'now'),
      "success": random.choice(
          [True, False]
      )
  }

  logs.append(log)

  df4=pd.DataFrame(logs)
```

```python
# Create a large dataset to predict student chances of graduation

data = []

for i in range(50):
  student = {
      "student_id": i + 1,
      "age": round(random.uniform(1,17), 0),
      "gender": random.choice(["Male", "Female"]),
      "socio_economic_score": random.choice(["Poor",
                                             "Lower_class",
                                             "Upper_middle_class",
                                             "Upper_class",
                                             "Wealthy"]),
      "attendance_rate": round(random.uniform(0, 100), 2),
      "average_assignment_score": random.randint(0,100),
      "midterm_exam_score": random.randint(0, 100),
      "final_exam_score": random.randint(0, 100),
      "study_hours_per_week": random.randint(0, 100),
      "lms_logins_per_week": random.randint(0, 100),
      "part_time_job_hours": random.randint(0, 100),
      "financial_aid": random.choice(["True", "False"]),
      "disciplinary_record": random.choice(
          [
          "None",
          "Moderate",
          "High"
      ]
          ),
      "graduation_outcome": random.choice(
          [
          "On_Time",
          "Delayed",
          "Dropout_Risk",
          "High_Achiever"
      ]
          )

  }

  data.append(student)

  df_=pd.DataFrame(data)
```
